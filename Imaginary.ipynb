```python
%%manim -pqh -r 1080,1920 ImaginaryNumbersAreReal
Text.set_default(font="Avenir")


from manim import *
from MF_Tools import * 
class ImaginaryNumbersAreReal(Scene):
    def construct(self):
        # HOOK: "REAL" sliced by i
        real_text = Text("REAL", font_size=144)
        i_sword = Text("i", font_size=72, color=BLUE).rotate(PI/4)
        i_sword.next_to(real_text, UP, buff=0).shift(LEFT)

        self.play(Write(real_text), FadeIn(i_sword))
        self.wait(0.2)

        # Slash animation
        slash_line = Line(LEFT * 2, RIGHT * 2, color=BLUE).rotate(PI/4)
        self.play(GrowFromCenter(slash_line), run_time=0.3)
        self.remove(slash_line)

        # Split "REAL" and add glitch
        left_half = Text("RE", font_size=144).shift(LEFT * 0.5)
        right_half = Text("AL", font_size=144).shift(RIGHT * 0.5)
        self.play(
            FadeOut(real_text),
            FadeIn(left_half, shift=LEFT * 0.2),
            FadeIn(right_half, shift=RIGHT * 0.2),
            i_sword.animate.shift(DOWN * 0.5 + RIGHT * 0.3).rotate(-PI/6),
        )
        self.wait(0.2)
        # CAT REACTION (image)
        cat = ImageMobject("cat2.png").scale(0.3).next_to(left_half, DOWN)
        question = Text("???", font_size=64).next_to(cat, RIGHT)
        self.play(FadeIn(cat), FadeIn(question))
        self.wait(0.3)
        # GLITCH
        self.play(
            cat.animate.scale(1.2),
            question.animate.shift(UP * 0.1),
            Flash(cat.get_center(), color=RED),
            run_time=0.2
        )

        # Zoom into cat's head (cut to brain scene)
        self.play(FadeOut(VGroup(left_half, right_half, i_sword, question)))
        brain_bg = Rectangle(width=7, height=4, color=WHITE, fill_opacity=0.1)
        formulae = VGroup(
            MathTex("i^2 = -1"),
            MathTex("e^{i\pi} + 1 = 0"),
        ).arrange(DOWN, buff=0.4)

        self.play(FadeIn(brain_bg), Write(formulae))
        self.wait(0.5)

        # Complex plane
        self.play(FadeOut(formulae), FadeOut(brain_bg))
        plane = ComplexPlane().add_coordinates()
        self.play(Create(plane))

        # Cat gets yeeted onto the plane
        cat.move_to(LEFT * 3 + DOWN)
        self.play(cat.animate.move_to(ORIGIN).scale(0.8).rotate(PI/4), run_time=0.6)

        # EXPLANATION: cat rotates because multiplying by i = 90° rotation
        explanation = Text("Multiplying by i = 90° rotation", font_size=36).next_to(plane, UP)
        self.play(Write(explanation), run_time=0.5)
        self.play(Rotate(cat, angle=PI/2, run_time=0.4))
        self.wait(0.3)
        self.play(FadeOut(explanation))

        # Final frame: cat holds sign
        sign = Text("i = Real (deal with it)", font_size=48).next_to(cat, UP)
        self.play(Write(sign), run_time=0.5)
        self.wait(1)

        # Fade out
        self.play(FadeOut(VGroup(sign, plane)),FadeOut(cat))

```
