```python
%%manim -pqh -r 1080,1920 GraphTransformationsTikTok

from manim import *
Text.set_default(font="Avenir")

class GraphTransformationsTikTok(Scene):
    def construct(self):
        def add_highlight(mobj, color):
            return SurroundingRectangle(mobj, color=color, buff=0.2)

        def add_catcam():
            return ImageMobject("cat2.png").scale(0.4).to_corner(DR, buff=0.5)

       
        # Main Hook Text
        hook = Text("A-Level Maths: Brainrot Edition", font_size=54, color=RED)
        hook.scale(0.1).move_to(UP * 2)  # start tiny at top

        # Animate like a cursed TikTok title drop
        self.play(
            hook.animate.scale(10).move_to(ORIGIN),
            rate_func=there_and_back_with_pause,
            run_time=1.4
        )
        self.wait(0.3)

        # Subheading Episode
        subtitle = Text("Ep 2: Graphs go up, vibes go down", font_size=42, color=WHITE).next_to(hook, DOWN)
        self.play(FadeIn(subtitle, shift=UP), run_time=0.5)
        self.wait(1.3)

        # Quick fade out
        self.play(FadeOut(hook), FadeOut(subtitle))
        # === Axes Setup ===
        axes = Axes(
            x_range=[-5, 5, 1],
            y_range=[-5, 5, 1],
            tips=False,
            axis_config={"include_numbers": True}
        )
        labels = axes.get_axis_labels(x_label="x", y_label="f(x)")
        labels[1].shift(DOWN * 0.3)

        base_func_orig = axes.plot(lambda x: x**2, color=BLUE)
        base_label = MathTex("f(x) = x^2").scale(1.2).to_corner(UL).shift(DOWN * 0.3)

        self.play(Create(axes), Write(labels))
        base_func = base_func_orig.copy()
        self.play(Create(base_func), Write(base_label))
        self.wait(1)

        # === f(x) + 2 ===
        fx_plus_2 = axes.plot(lambda x: x**2 + 2, color=GREEN)
        plus_2_label = MathTex("f(x) + 2").scale(1.5).set_color(GREEN).move_to(UP * 4)
        plus_2_box = add_highlight(plus_2_label, GREEN)
        up_arrow = Arrow(start=DOWN, end=UP, color=GREEN).next_to(axes.c2p(0, 2), LEFT)
        up_text = Text("Move UP 2 units", font_size=36).set_color(GREEN).to_edge(DOWN).shift(DOWN * 0.2)

        catcam = add_catcam()
        self.add(catcam)

        self.play(
            Transform(base_func, fx_plus_2),
            Write(plus_2_label), Create(plus_2_box),
            GrowArrow(up_arrow),
            FadeIn(up_text),
            catcam.animate.shift(UP * 0.5)
        )
        self.wait(1)
        self.play(*map(FadeOut, [plus_2_label, plus_2_box, up_arrow, up_text, base_func, catcam]))

        # === Reset ===
        base_func = base_func_orig.copy()
        self.add(base_func)

        # === f(x - 2) ===
        fx_minus_2 = axes.plot(lambda x: (x - 2)**2, color=YELLOW)
        minus_2_label = MathTex("f(x - 2)").scale(1.5).set_color(YELLOW).move_to(UP * 4)
        minus_2_box = add_highlight(minus_2_label, YELLOW)
        right_arrow = Arrow(start=LEFT, end=RIGHT, color=YELLOW).next_to(axes.c2p(2, 0), DOWN)
        right_text = Text("Move RIGHT 2 units 😮", font_size=36).set_color(YELLOW).to_edge(DOWN).shift(DOWN * 0.2)

        catcam = add_catcam()
        self.add(catcam)

        self.play(
            Transform(base_func, fx_minus_2),
            Write(minus_2_label), Create(minus_2_box),
            GrowArrow(right_arrow),
            FadeIn(right_text),
            catcam.animate.shift(RIGHT * 0.5)
        )
        self.wait(1)
        self.play(*map(FadeOut, [minus_2_label, minus_2_box, right_arrow, right_text, base_func, catcam]))

        # === Reset ===
        base_func = base_func_orig.copy()
        self.add(base_func)

        # === -f(x) ===
        neg_fx = axes.plot(lambda x: -x**2, color=RED)
        neg_fx_label = MathTex("-f(x)").scale(1.5).set_color(RED).move_to(UP * 4)
        neg_fx_box = add_highlight(neg_fx_label, RED)
        x_reflect_arrow = CurvedArrow(start_point=axes.c2p(-3, 3), end_point=axes.c2p(-3, -3), color=RED)
        reflect_text = Text("REFLECT in x-axis 🔥", font_size=36).set_color(RED).to_edge(DOWN).shift(DOWN * 0.2)

        catcam = add_catcam()
        self.add(catcam)

        self.play(
            Transform(base_func, neg_fx),
            Write(neg_fx_label), Create(neg_fx_box),
            Create(x_reflect_arrow),
            FadeIn(reflect_text),
            catcam.animate.rotate(PI)  # flip upside down
        )
        self.wait(1)
        self.play(*map(FadeOut, [neg_fx_label, neg_fx_box, x_reflect_arrow, reflect_text, base_func, catcam]))

        # === Reset ===
        base_func = base_func_orig.copy()
        self.add(base_func)

        # === f(-x) ===
        reflect_y = axes.plot(lambda x: (-x)**2, color=PURPLE)
        reflect_y_label = MathTex("f(-x)").scale(1.5).set_color(PURPLE).move_to(UP * 4)
        reflect_y_box = add_highlight(reflect_y_label, PURPLE)
        y_reflect_arrow = CurvedArrow(start_point=axes.c2p(3, -3), end_point=axes.c2p(-3, -3), color=PURPLE)
        reflect_y_text = Text("REFLECT in y-axis???", font_size=36).set_color(PURPLE).to_edge(DOWN).shift(DOWN * 0.2)

        catcam = add_catcam()
        self.add(catcam)

        self.play(
            Transform(base_func, reflect_y),
            Write(reflect_y_label), Create(reflect_y_box),
            Create(y_reflect_arrow),
            FadeIn(reflect_y_text),
            catcam.animate.flip(LEFT)
        )
        self.wait(1)
        self.play(*map(FadeOut, [reflect_y_label, reflect_y_box, y_reflect_arrow, reflect_y_text, base_func, catcam]))

        # === Summary ===
        self.play(
            FadeOut(axes),
            FadeOut(labels),
            FadeOut(base_label)
        )

        # Bullet list with bigger size
        bullets = BulletedList(
            "f(x) + a → move up/down",
            "f(x + a) → move left/right",
            "-f(x) → reflect in x-axis",
            "f(-x) → reflect in y-axis",
            font_size=38,  # increased size
        )
        bullets.set_color_by_tex_to_color_map({
            "f(x) + a": GREEN,
            "f(x + a)": YELLOW,
            "-f(x)": RED,
            "f(-x)": PURPLE
        })

        cheat_title = Text("Graph Transformations Cheat Sheet", font_size=48, weight=BOLD)
        cheat_title.next_to(bullets, UP, buff=0.6)

        cheat_sheet = VGroup(cheat_title, bullets)
        cheat_sheet.move_to(ORIGIN)

        # Background panel behind the group (slightly bigger than the group)
        bg_panel = RoundedRectangle(
            corner_radius=0.3, 
            width=cheat_sheet.width + 1, 
            height=cheat_sheet.height + 1, 
            color=WHITE, 
            fill_opacity=0.15
        ).move_to(cheat_sheet)

        # Animate in
        self.play(
            FadeIn(bg_panel),
            FadeIn(cheat_sheet, shift=UP),
            run_time=1.2
        )
        self.wait(3)

        # Outro tease
        self.play(*[FadeOut(mob) for mob in self.mobjects])
        next_up = Text("Next Time: Why Is e So Obsessed with Growth? 📈", font_size=42, color=BLUE)
        self.play(Write(next_up, run_time=0.6))
        self.wait(2)



```
