```python
%%manim -pqh -r 1080,1920 DivergentSeriesEqualsNegativeOneTwelfth
from manim import *
from MF_Tools import *

Text.set_default(font="Avenir")

class DivergentSeriesEqualsNegativeOneTwelfth(MovingCameraScene):
    def construct(self):
        # Hook: Classic divergent series
        hook = MathTex("1 + 2 + 3 + 4 + \\dots")
        hook.scale(1.5).set_color(RED)
        self.play(Write(hook))
        self.wait(0.3)

        q = Text("What if I told you this sum equals... -1/12?!", font_size=36)
        q.next_to(hook, DOWN)
        self.play(FadeIn(q))
        self.wait(0.6)
        self.play(FadeOut(hook), FadeOut(q))
        self.wait(0.3)

        # Step 1: Define zeta function
        zeta_def = MathTex(r"\zeta(s) = \sum_{n=1}^\infty \frac{1}{n^s}")
        zeta_def.scale(1.2)
        self.play(Write(zeta_def))
        self.wait(0.7)

        # Step 2: Plug in s = -1
        plug = MathTex(r"\zeta(-1) = 1 + 2 + 3 + 4 + \dots")
        plug.scale(1.2)
        self.play(Transform(zeta_def, plug))
        self.wait(0.7)

        # Step 3: Zeta at negative integers via Bernoulli numbers
        zeta_neg = MathTex(r"\zeta(-1) = -\frac{B_2}{2}")
        zeta_neg.scale(1.2)
        self.play(TransformMatchingTex(zeta_def, zeta_neg))
        self.wait(0.7)

        # Step 4: Show the generating function
        gen_func = MathTex(r"\frac{x}{e^x - 1} = \sum_{n=0}^\infty \frac{B_n}{n!} x^n")
        gen_func.scale(0.95)
        gen_func.set_color(YELLOW)
        gen_func.next_to(zeta_neg,DOWN)
        self.play(Write(gen_func))
        self.wait(1)

        # Step 5: Expand up to x^2
        taylor = MathTex(
            r"\frac{x}{e^x - 1} = 1 - \frac{x}{2} + \frac{x^2}{12} - \frac{x^4}{720} + \dots"
        )
        taylor.scale(0.95)
        taylor.next_to(gen_func, DOWN)
        self.play(Write(taylor))
        self.wait(1)

        # Step 6: Match coefficients
        match = MathTex(r"\text{So,} ( B_2 = \frac{1}{6})", font_size=36)
        match.set_color(WHITE)
        match.next_to(taylor, DOWN)
        self.play(FadeIn(match))
        self.wait(0.8)

        # Step 7: Plug back into zeta(-1)
        b_plug = MathTex(r"\Rightarrow \zeta(-1) = -\frac{1}{12}")
        b_plug.scale(1.4)
        b_plug.set_color(GREEN)
        self.play(FadeOut(zeta_neg), FadeOut(gen_func), FadeOut(taylor), FadeOut(match), Write(b_plug))
        self.wait(1)

        # Step 8: Show final equivalence
        final = MathTex("1 + 2 + 3 + 4 + \\dots = -\\frac{1}{12}")
        final.scale(1.5)
        final.set_color(GREEN)
        self.play(FadeOut(b_plug),Write(final))
        self.wait(1)

        # Optional cheeky cat comment
        cat = Text("don’t try this on your A-Levels.", font_size=34, slant=ITALIC)
        cat.set_color(YELLOW)
        cat.next_to(final, DOWN)
        self.play(FadeIn(cat))
        self.wait(1)

        # Pulse
        pulse = Circle(radius=1.2, color=YELLOW, stroke_width=2)
        pulse.move_to(final)
        self.play(GrowFromCenter(pulse), FadeOut(pulse, scale=2), run_time=0.5)

        self.play(FadeOut(final), FadeOut(cat))
```
