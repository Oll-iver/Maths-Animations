```python
%%manim -pqh -r 1080,1920 WhatIsAFunction

from manim import *
Text.set_default(font="Avenir")

class WhatIsAFunction(Scene):
    def construct(self):
        # Main Hook Text
        hook = Text("A-Level Maths: Brainrot Edition", font_size=54, color=RED)
        hook.scale(0.1).move_to(UP * 2)  # start tiny at top

        # Animate like a cursed TikTok title drop
        self.play(
            hook.animate.scale(10).move_to(ORIGIN),
            rate_func=there_and_back_with_pause,
            run_time=1.4
        )
        self.wait(0.3)

        # Subheading Episode
        subtitle = Text("Ep 1: This is NOT a Function", font_size=42, color=WHITE).next_to(hook, DOWN)
        self.play(FadeIn(subtitle, shift=UP), run_time=0.5)
        self.wait(1.3)

        # Quick fade out
        self.play(FadeOut(hook), FadeOut(subtitle))

        # Domain & Codomain Circles
        domain = Circle(radius=1.5, color=BLUE).shift(LEFT * 3)
        codomain = Circle(radius=1.5, color=GREEN).shift(RIGHT * 3)
        domain_label = Text("Inputs", font_size=24).next_to(domain, UP)
        codomain_label = Text("Outputs", font_size=24).next_to(codomain, UP)

        # Numbers
        inputs = VGroup(*[Text(str(n)) for n in range(1, 4)])
        outputs = VGroup(*[Text(str(n**2)) for n in range(1, 4)])

        for i, item in enumerate(inputs):
            item.move_to(domain.get_center() + DOWN * (i - 1))
        for i, item in enumerate(outputs):
            item.move_to(codomain.get_center() + DOWN * (i - 1))

        # Arrows
        arrows = VGroup()
        for i in range(3):
            arrow = Arrow(
                start=inputs[i].get_right(),
                end=outputs[i].get_left(),
                buff=0.1,
                stroke_width=3
            )
            arrows.add(arrow)

        self.play(GrowFromCenter(domain), GrowFromCenter(codomain), run_time=0.6)
        self.play(FadeIn(domain_label, shift=UP), FadeIn(codomain_label, shift=UP), run_time=0.4)
        for inp, out in zip(inputs, outputs):
            self.play(FadeIn(inp, shift=LEFT), FadeIn(out, shift=RIGHT), run_time=0.3)
        for arrow in arrows:
            self.play(GrowArrow(arrow), run_time=0.25)

        self.wait(1)

        # Function machine metaphor
        self.play(*[FadeOut(mob) for mob in self.mobjects])
        machine = Rectangle(width=4, height=2, color=YELLOW).shift(DOWN)
        f_text = Text("f(x) = x²", font_size=36).move_to(machine.get_center())
        f_text.set_opacity(0)
        input_val = Text("3", font_size=36).shift(LEFT * 5)
        output_val = Text("9", font_size=36).shift(RIGHT * 5)
        input_arrow = Arrow(start=input_val.get_right(), end=machine.get_left(), buff=0.1)
        output_arrow = Arrow(start=machine.get_right(), end=output_val.get_left(), buff=0.1)

        self.play(Create(machine), run_time=0.5)
        self.play(f_text.animate.set_opacity(1), run_time=0.5)
        self.play(Write(input_val), Create(input_arrow))
        self.play(Write(output_val), Create(output_arrow))
        self.wait(1.5)

        # Show a non-function (multi-output)
        self.play(*[FadeOut(mob) for mob in self.mobjects])
        machine = Rectangle(width=4, height=2, color=YELLOW).shift(UP*1.55+LEFT*0.7)
        bad_function = MathTex("f(x) = x \\pm 4").move_to(machine.get_center())
        bad_input = Text("5", font_size=36).shift(LEFT * 3)
        bad_output1 = Text("1", font_size=36).shift(RIGHT * 2 + UP * 0.5)
        bad_output2 = Text("9", font_size=36).shift(RIGHT * 2 + DOWN * 0.5)
        bad_arrow1 = Arrow(start=bad_input.get_right(), end=bad_output1.get_left(), buff=0.1, color=RED)
        bad_arrow2 = Arrow(start=bad_input.get_right(), end=bad_output2.get_left(), buff=0.1, color=RED)
        cross = Cross(bad_input, color=RED).scale(1.2)
        cross.set_stroke(width=6)


        self.play(Create(machine),Write(bad_function),Write(bad_input), Write(bad_output1), Write(bad_output2))
        self.play(Flash(bad_arrow1.get_start(), color=RED, flash_radius=0.4), run_time=0.2)
        self.play(Create(bad_arrow1), Create(bad_arrow2), run_time=0.4)
        self.wait(0.5)
        self.play(Create(cross))
        self.wait(1)

        # Outro Tease
        self.play(*[FadeOut(mob) for mob in self.mobjects])
        next_up = Text("Next Time: When Graphs Go Rogue 📈", font_size=42, color=BLUE)
        self.play(Write(next_up, run_time=0.6))
        self.wait(2)

```
