```python
%%manim -p KochSnowflakeIterativeRevealFinal
from manim import *
import numpy as np

def koch_points(points, depth):
    if depth == 0:
        return points
    new_points = []
    for i in range(len(points) - 1):
        A = points[i]
        B = points[i+1]
        # Divide segment into thirds
        one_third = A + (B - A) / 3
        two_third = A + 2 * (B - A) / 3
        # Create the peak by rotating the middle segment by 60°
        peak = one_third + rotate_vector(two_third - one_third, -1*PI/3)
        new_points.extend([A, one_third, peak, two_third])
    new_points.append(points[-1])
    return koch_points(new_points, depth - 1)

class KochSnowflakeIterativeRevealFinal(MovingCameraScene):
    def construct(self):
        # Define the base equilateral triangle.
        side_length = 3
        A = np.array([-side_length/2, 0, 0])
        B = np.array([side_length/2, 0, 0])
        C = np.array([0, side_length * np.sqrt(3)/2, 0])
        base_triangle = [A, B, C, A]
        # Number of iterations (depth) to reveal.
        max_depth = 6
        # Generate a list of snowflake VMobjects for each iteration.
        snowflakes = []
        for d in range(max_depth + 1):
            pts = koch_points(base_triangle, d)
            snowflake = VMobject()
            snowflake.set_points_as_corners(pts)
            snowflake.set_stroke(color=BLUE, width=2)
            snowflakes.append(snowflake)

        # Create a label for the iteration counter, starting with iteration 0.
        iteration_label = Text("Iteration 0", font_size=36)
        iteration_label.to_edge(UP)
        self.play(Create(iteration_label))

        # Start by showing the base triangle.
        self.play(Create(snowflakes[0]))
        self.wait(1)
        
        # Iteratively transform the snowflake through higher recursion levels.
        for i in range(1, len(snowflakes)):
            new_label=Text(f"Iteration {i}",font_size=36).to_edge(UP)
            self.play(Transform(iteration_label,new_label),run_time=0.5)
            if i==1:
                target_center = snowflakes[i].get_center()
                target_width = snowflakes[i].get_width() * 3.5
                self.play(
                    self.camera.frame.animate.move_to(target_center).set(width=target_width),
                    run_time=1
                )
            self.play(FadeOut(snowflakes[i-1]), FadeIn(snowflakes[i]), run_time=2)
            self.wait(1)
        # Keep the final iteration on screen for a moment.
        self.wait(1)

        final_center = snowflakes[-1].get_center() + UP * 1.5
        target_width = snowflakes[-1].get_width() * 0.5  # Adjust scale as needed

        self.play(
            self.camera.frame.animate.move_to(final_center).set(width=target_width),
            run_time=1
        )

        self.wait(2)

```
