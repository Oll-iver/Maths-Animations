```python
%%manim -pqh -r 1080,1920 CoordinatesAreALie
from manim import *
import numpy as np

class CoordinatesAreALie(MovingCameraScene):
    def construct(self):
        # === SETUP ===
        grid = NumberPlane(
            x_range=[-6, 6, 1],
            y_range=[-4, 4, 1],
            background_line_style={
                "stroke_color": BLUE_E,
                "stroke_width": 1,
                "stroke_opacity": 0.5
            }
        )
        self.add(grid)

        # Cat at a fixed point
        cat_pos = np.array([3, 2, 0])
        cat = ImageMobject("cat2.png").scale(0.5).move_to(cat_pos)

        # Standard basis vectors
        i_hat = Arrow(ORIGIN, RIGHT, buff=0, color=RED)
        j_hat = Arrow(ORIGIN, UP, buff=0, color=GREEN)

        self.play(FadeIn(cat), GrowArrow(i_hat), GrowArrow(j_hat))

        # Intro Text
        text1 = Text("This cat is a point in space.", font="Futura").scale(0.6).to_edge(UP)
        text2 = Text("Its coordinates are (3, 2).", font="Futura").scale(0.6).next_to(text1, DOWN)
        self.play(Write(text1), Write(text2))
        self.wait(0.8)

        self.play(FadeOut(text1), FadeOut(text2))

        # === NEW BASIS ===
        basis_angle = PI / 6  # 30 degrees
        b1 = rotate_vector(RIGHT, basis_angle)
        b2 = rotate_vector(UP, basis_angle)
        b1_arrow = Arrow(ORIGIN, b1, buff=0, color=YELLOW)
        b2_arrow = Arrow(ORIGIN, b2, buff=0, color=ORANGE)

        # Label new basis
        label_b1 = MathTex(r"\vec{b}_1").next_to(b1_arrow, DOWN)
        label_b2 = MathTex(r"\vec{b}_2").next_to(b2_arrow, LEFT)

        new_basis_text = Text("But what if we used these directions?", font="Futura").scale(0.6).to_edge(UP)
        self.play(Write(new_basis_text))
        self.play(GrowArrow(b1_arrow), GrowArrow(b2_arrow), FadeIn(label_b1), FadeIn(label_b2),FadeOut(i_hat),FadeOut(j_hat))
        self.wait(0.5)

        self.play(FadeOut(new_basis_text))

        # === DECOMPOSE CAT IN NEW BASIS ===
        B = np.column_stack([b1[:2], b2[:2]])  # 2x2 matrix
        new_coords = np.linalg.solve(B, cat_pos[:2])
        coeffs_text = Text(f"New coordinates: ({new_coords[0]:.2f}, {new_coords[1]:.2f})", font="Futura").scale(0.55).to_edge(UP)

        # Show scaled b1 and b2
        end_b1 = b1 * new_coords[0]
        end_b2 = cat_pos - np.append(end_b1[:2], 0)  # Make sure it's 3D
        b1_scaled = Arrow(ORIGIN, end_b1, buff=0, color=YELLOW_D)
        b2_scaled = Arrow(end_b1, end_b1 + end_b2, buff=0, color=ORANGE)


        self.play(GrowArrow(b1_scaled))
        self.play(GrowArrow(b2_scaled))
        self.wait(0.3)
        self.play(Write(coeffs_text))
        self.wait(1.0)

        # === ROTATE BASIS + SHOW COORD CHANGE ===
        self.play(FadeOut(coeffs_text, b1_scaled, b2_scaled))
        rotate_text = Text("Change the basis... change the coordinates.", font="Futura").scale(0.6).to_edge(UP)
        self.play(Write(rotate_text))

        def update_basis(group, alpha):
            angle = basis_angle + alpha * PI/4
            new_b1 = rotate_vector(RIGHT, angle)
            new_b2 = rotate_vector(UP, angle)
            B_new = np.column_stack([new_b1[:2], new_b2[:2]])
            coords = np.linalg.solve(B_new, cat_pos[:2])

            group[0].put_start_and_end_on(ORIGIN, new_b1)
            group[1].put_start_and_end_on(ORIGIN, new_b2)
            group[2].next_to(group[0], DOWN)
            group[3].next_to(group[1], LEFT)
            group[4].become(Text(f"({coords[0]:.2f}, {coords[1]:.2f})", font="Futura").scale(0.5).next_to(cat, UP))

        # Create rotating versions directly from originals
        b1_copy = b1_arrow
        b2_copy = b2_arrow
        label1 = label_b1
        label2 = label_b2
        coord_display = Text(f"({new_coords[0]:.2f}, {new_coords[1]:.2f})", font="Futura").scale(0.5).next_to(cat, UP)

        basis_group = VGroup(b1_copy, b2_copy, label1, label2, coord_display)
        self.add(basis_group)
        self.play(FadeIn(coord_display))
        self.wait(0.3)
        self.play(UpdateFromAlphaFunc(basis_group, update_basis, run_time=4))
        self.wait(0.5)
        self.play(FadeOut(rotate_text))
        self.play(FadeOut(basis_group))

        # === PUNCHLINE ===
        punch = Text("The cat doesn’t care what basis you choose.", font="Futura").scale(0.55).to_edge(UP)
        self.play(Write(punch))
        self.wait(1.5)
        outro = Text("It knows where it is.", font="Futura").scale(0.55).next_to(punch, DOWN)
        self.play(Write(outro))
        self.wait(1.5)

        final_cta = Text("Follow for more geometry in disguise.", font="Futura").scale(0.5).to_edge(DOWN).shift(UP*3)
        self.play(FadeIn(final_cta))
        self.wait(2)

```
