```python
%%manim -pqh -r 1080,1920 GraphTypes

from manim import *
import numpy as np
Text.set_default(font="Avenir")

class GraphTypes(Scene):
    def construct(self):
        def add_highlight(mobj, color):
            return SurroundingRectangle(mobj, color=color, buff=0.2)

        def add_catcam():
            return ImageMobject("cat2.png").scale(0.4).to_corner(DR, buff=0.5)

        # Hook
        hook = Text("A-Level Maths: Brainrot Edition", font_size=54, color=RED)
        hook.scale(0.1).move_to(UP * 2)
        self.play(hook.animate.scale(10).move_to(ORIGIN), rate_func=there_and_back_with_pause, run_time=1.4)
        self.wait(0.3)

        subtitle = Text("Ep 4: Linear, Exponential, Parabolic and Hyperbolic", font_size=42, color=WHITE).next_to(hook, DOWN)
        self.play(FadeIn(subtitle, shift=UP), run_time=0.5)
        self.wait(1.3)
        self.play(FadeOut(hook), FadeOut(subtitle))

        cat = add_catcam()
        self.add(cat)

        axes = Axes(
            x_range=[-4, 4],
            y_range=[-4, 8],
            axis_config={"include_tip": False}
        ).scale(1.1).shift(DOWN * 0.5)
        labels = axes.get_axis_labels(x_label="x", y_label="y")
        self.play(Create(axes), Write(labels))

        # === Linear ===
        linear = axes.plot(lambda x: x + 1, color=BLUE)
        linear_label = MathTex("y = x + 1").scale(1.2).set_color(WHITE).to_corner(UL).shift(DOWN*0.5)
        linear_text = Text("Linear: Steady growth. No surprises.", font_size=36, color=BLUE).to_edge(DOWN)
        self.play(Create(linear), FadeIn(linear_label))
        self.play(Write(linear_text), cat.animate.shift(RIGHT*0.2))
        self.wait(1)
        self.play(*map(FadeOut, [linear_text, linear_label, linear]))

        # === Exponential ===
        exp = axes.plot(lambda x: np.exp(x - 2), color=YELLOW)
        exp_label = MathTex("y = e^{x-2}").scale(1.2).set_color(WHITE).to_corner(UL).shift(DOWN*0.5)
        asymptote = DashedLine(start=axes.c2p(-4, 0), end=axes.c2p(4, 0), color=YELLOW)
        exp_text = Text("Exponential: Starts slow. Ends fast.", font_size=36, color=YELLOW).to_edge(DOWN)
        self.play(Create(exp), FadeIn(exp_label), Create(asymptote))
        self.play(Write(exp_text), cat.animate.shift(UP * 0.3))
        self.wait(1)
        self.play(*map(FadeOut, [exp_text, exp_label, asymptote, exp]))

        # === Parabolic ===
        parabola = axes.plot(lambda x: x**2 - 2, color=GREEN)
        parabola_label = MathTex("y = x^2 - 2").scale(1.2).set_color(WHITE).to_corner(UL).shift(DOWN*0.5)
        para_text = Text("Parabolic: Dramatic arc. Symmetric.", font_size=36, color=GREEN).to_edge(DOWN)
        self.play(Create(parabola), FadeIn(parabola_label))
        self.play(Write(para_text), cat.animate.flip(LEFT))
        self.wait(1)
        self.play(*map(FadeOut, [para_text, parabola_label, parabola]))

        # === Hyperbolic ===
        hyperbola = axes.plot(lambda x: 4 / (x + 0.01), color=RED)
        hyperbola_label = MathTex("y = \\frac{4}{x}").scale(1.2).set_color(WHITE).to_corner(UL).shift(DOWN*0.5)
        vertical_asymp = DashedLine(start=axes.c2p(0, -4), end=axes.c2p(0, 8), color=RED)
        horizontal_asymp = DashedLine(start=axes.c2p(-4, 0), end=axes.c2p(4, 0), color=RED)
        hyper_text = Text("Hyperbolic: Gets close... but extends to infinity ", font_size=34, color=WHITE).to_edge(DOWN)
        self.play(Create(hyperbola), FadeIn(hyperbola_label), Create(vertical_asymp), Create(horizontal_asymp))
        self.play(Write(hyper_text), cat.animate.rotate(PI/2))
        self.wait(1.2)
        self.play(*map(FadeOut, [hyper_text, hyperbola_label, vertical_asymp, horizontal_asymp, hyperbola]))

        # Fade out all
        self.play(FadeOut(axes), FadeOut(labels), FadeOut(cat))

        # === Final Comparison Text ===
        summary = BulletedList(
            "Linear: Constant rate",
            "Exponential: Growth monster",
            "Parabolic: Symmetric drama",
            "Hyperbolic: Asymptotic chaos",
            font_size=36
        )
        summary.set_color_by_tex_to_color_map({
            "Linear": BLUE,
            "Exponential": YELLOW,
            "Parabolic": GREEN,
            "Hyperbolic": RED
        })

        summary_title = Text("The Final Verdict", font_size=46).next_to(summary, UP, buff=0.6)
        bg_panel = RoundedRectangle(corner_radius=0.3, width=summary.width + 1, height=summary.height + 1, fill_opacity=0.15).move_to(summary)

        self.play(FadeIn(bg_panel), FadeIn(summary_title), FadeIn(summary))
        self.wait(3)

        self.play(*[FadeOut(m) for m in self.mobjects])
        outro = Text("Next Time: An intro to integrals", font_size=42, color=BLUE)
        self.play(Write(outro))
        self.wait(2)
```
