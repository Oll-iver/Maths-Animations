```python
%%manim -pqh -r 1080,1920 ExponentialsAreWeird
from manim import *
Text.set_default(font="Avenir")



class ExponentialsAreWeird(Scene):
    def construct(self):
        def add_highlight(mobj, color):
            return SurroundingRectangle(mobj, color=color, buff=0.2)

        def add_catcam():
            return ImageMobject("cat2.png").scale(0.4).to_corner(DR, buff=0.5)

        # TikTok Hook
        hook = Text("A-Level Maths: Brainrot Edition", font_size=54, color=RED)
        hook.scale(0.1).move_to(UP * 2)
        self.play(hook.animate.scale(10).move_to(ORIGIN), rate_func=there_and_back_with_pause, run_time=1.4)
        self.wait(0.3)

        subtitle = Text("Ep 3: e is NOT just a letter", font_size=42, color=WHITE).next_to(hook, DOWN)
        self.play(subtitle.animate.scale(1.2).shift(UP * 0.3), run_time=0.5)
        self.wait(1.3)

        self.play(FadeOut(hook), FadeOut(subtitle))

        # Add catcam
        cat = add_catcam()
        self.add(cat)

        # Set up axes

        
        axes = Axes(
            x_range=[-2, 3],
            y_range=[0, 10],
            axis_config={"include_tip": False}
        ).scale(1.2)
        axes.move_to(DOWN * 0.5)
        labels = axes.get_axis_labels(x_label="x", y_label="y")
        self.play(Create(axes), FadeIn(labels))

        # Show exponential function
        exp_graph = axes.plot(lambda x: np.exp(x), color=YELLOW)
        exp_label = MathTex(r"y = e^x").next_to(exp_graph, RIGHT).shift(UP*0.5)
        self.play(Create(exp_graph), FadeIn(exp_label))

        # Add highlight around key point
        point = Dot(axes.coords_to_point(0, 1), color=WHITE)
        point_label = MathTex(r"(0, 1)").next_to(point, UL)
        highlight = add_highlight(point_label, BLUE)
        self.play(FadeIn(point), FadeIn(point_label), Create(highlight))
        self.wait(0.5)

        # Explain tangent at x=0 is 1
        tangent = axes.plot(lambda x: x + 1, color=GREEN)
        tangent_label = MathTex(r"\text{slope at } x=0 \text{ is } 1").to_edge(UP).shift(RIGHT*0.7,DOWN)
        self.play(Create(tangent), FadeIn(tangent_label))
        self.wait(1.0)

        # Position reference for text block (left center of graph)
        text_pos = axes.coords_to_point(-0.5, 4.5)

        # Insert: Derivative of e^x
        deriv = MathTex(r"\frac{d}{dx}(e^x) = e^x", color=BLUE_C).move_to(text_pos).shift(LEFT*0.5)
        deriv_text = Text("It’s its own derivative.", font_size=30).next_to(deriv, UP)
        self.play(Write(deriv_text), Write(deriv))
        self.wait(1.5)
        self.play(FadeOut(deriv_text), FadeOut(deriv))

        # Insert: Integral of e^x
        integral = MathTex(r"\int e^x\,dx = e^x + c", font_size=60, color=GREEN_C).move_to(text_pos).shift(LEFT*0.4)
        integral_text = Text("Also its own integral. Chill like that.", font_size=30).next_to(integral, UP)
        self.play(Write(integral_text), Write(integral))
        self.wait(1.5)
        self.play(FadeOut(integral_text), FadeOut(integral))

        # Insert: limit definition of e
        limit_eq = MathTex(r"e = \lim_{n \to \infty} \left(1 + \dfrac{1}{n}\right)^n", color=PURPLE_C).move_to(text_pos).shift(LEFT*0.8)
        limit_text = Text("e is literally born from infinity", font_size=30).next_to(limit_eq, UP)
        self.play(Write(limit_text), Write(limit_eq))
        self.wait(1.5)
        self.play(FadeOut(limit_text), FadeOut(limit_eq))

        # Emphasise the weirdness
        weird_text = Text("It’s the only function that copies itself.", font_size=32, color=ORANGE).move_to(text_pos)
        self.play(Write(weird_text))
        self.wait(1.2)

        outro = Text("Maths is just weird like that.", font_size=36).next_to(weird_text, UP)
        self.play(Write(outro))
        self.wait(1.5)

        # Spiral into existential dread
        dread = Text("do numbers even mean anything?", font_size=30, color=GRAY).to_corner(UL)
        self.play(SpinInFromNothing(dread))
        self.wait(1.2)

         # Clean fade out of everything first
        self.play(*[FadeOut(mob) for mob in self.mobjects])

        # True to the brainrot outro style
        next_up = Text("Next time - UFC 318: Linear vs Exponential", font_size=42, color=BLUE)
        self.play(Write(next_up, run_time=0.6))
        self.wait(2)
```
