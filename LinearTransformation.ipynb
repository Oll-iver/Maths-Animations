```python
%%manim -pqh -r 1080,1920 TikTokGridWarpWithCat
from manim import *

class TikTokGridWarpWithCat(MovingCameraScene):
    def construct(self):
        # === SETUP: Grid and Cat ===
        grid = NumberPlane(
            x_range=[-7, 7, 1],
            y_range=[-4, 4, 1],
            background_line_style={
                "stroke_color": BLUE_E,
                "stroke_width": 1,
                "stroke_opacity": 0.6
            }
        )
        cat = ImageMobject("cat2.png").scale(0.5).move_to([2, 1, 0])

        # Basis vectors (without labels)
        i_hat = Arrow(start=ORIGIN, end=RIGHT, buff=0, color=RED)
        j_hat = Arrow(start=ORIGIN, end=UP, buff=0, color=GREEN)

        # === PHASE 1: TikTok Visual Hook ===
        intro_text = Text("This is space.", font="Futura").scale(0.6).to_edge(UP)
        self.play(FadeIn(grid), FadeIn(cat), FadeIn(i_hat), FadeIn(j_hat), Write(intro_text), run_time=1.5)
        self.wait(0.3)

        warp_text = Text("And this is what math does to it.", font="Futura").scale(0.7).to_edge(UP)
        self.play(Transform(intro_text, warp_text), run_time=0.5)

        # Shear
        shear = [[1, 1], [0, 1]]
        self.play(
            grid.animate.apply_matrix(shear),
            cat.animate.apply_matrix(shear),
            i_hat.animate.apply_matrix(shear),
            j_hat.animate.apply_matrix(shear),
            run_time=1
        )
        self.wait(0.3)

        # Rotate
        angle = PI / 4
        rotate_matrix = [[np.cos(angle), -np.sin(angle)], [np.sin(angle), np.cos(angle)]]
        self.play(
            grid.animate.apply_matrix(rotate_matrix),
            cat.animate.apply_matrix(rotate_matrix),
            i_hat.animate.apply_matrix(rotate_matrix),
            j_hat.animate.apply_matrix(rotate_matrix),
            run_time=1
        )        # === TRANSITION: Reset Space ===
        reset_text = Text("Let’s reset space...", font="Futura").scale(0.6).to_edge(UP).shift(LEFT*2)
        self.play(FadeOut(intro_text), Write(reset_text))
        self.wait(0.5)

        # Reverse previous transformations (undo in reverse order)
        inverse_rotate = np.linalg.inv(rotate_matrix)
        self.play(
            grid.animate.apply_matrix(inverse_rotate),
            cat.animate.apply_matrix(inverse_rotate),
            i_hat.animate.apply_matrix(inverse_rotate),
            j_hat.animate.apply_matrix(inverse_rotate),
            run_time=1.5
        )
        inverse_shear1 = np.linalg.inv(shear)
        self.play(
            grid.animate.apply_matrix(inverse_shear1),
            cat.animate.apply_matrix(inverse_shear1),
            i_hat.animate.apply_matrix(inverse_shear1),
            j_hat.animate.apply_matrix(inverse_shear1),
            run_time=1.5
        )
        self.wait(0.3)

        # Swap text dramatically
        self.play(FadeOut(reset_text))
        boom_text = Text("Get ready for the most powerful matrix ever...", font="Futura").scale(0.6).to_edge(UP).shift(RIGHT*0.5)
        self.play(ScaleInPlace(boom_text,1.2))
        self.camera.frame.save_state()
        self.play(self.camera.frame.animate.scale(0.8).move_to(cat), run_time=1.2)

        self.wait(0.7)

        # Build up... then pop in matrix visually
        matrix_tex = MathTex(r"\begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}").scale(0.9).next_to(boom_text, DOWN).shift(RIGHT*0.1)
        self.play(FadeIn(matrix_tex), run_time=0.3)

        # Dramatic pause... then anticlimax caption
        anticlimax = Text("Start with the identity matrix — nothing moves.", font="Futura").scale(0.55).to_edge(UP).shift(RIGHT)
        self.play(Transform(boom_text, anticlimax), run_time=0.5)
        self.wait(0.3)

        # Apply identity (maybe with tiny 'shake' for fake tension)
        identity_matrix = [[1, 0], [0, 1]]
        self.play(
            grid.animate.apply_matrix(identity_matrix),
            cat.animate.apply_matrix(identity_matrix),
            i_hat.animate.apply_matrix(identity_matrix),
            j_hat.animate.apply_matrix(identity_matrix),
            run_time=1
        )

        # Bonus: wiggle the cat like “did it work?” (but nothing happened)
        self.play(cat.animate.shift(0.1 * RIGHT), run_time=0.2)
        self.play(cat.animate.shift(0.1 * LEFT), run_time=0.2)
        self.play(cat.animate.shift(0.05 * RIGHT), run_time=0.1)
        self.wait(0.5)

        # Sarcastic caption
        micro_text = Text("Yup. Nothing. That’s the point.", font="Futura").scale(0.45).next_to(boom_text, DOWN).shift(RIGHT*3.4)
        self.play(FadeIn(micro_text))
        self.play(Restore(self.camera.frame), run_time=0.5)        
        
        
        
        
        # === STRETCH TRANSFORMATION (TikTok-style) ===
        self.wait(0.3)
        stretch_text = Text("Okay... let's get weird.", font="Futura").scale(0.6).to_edge(UP)
        self.play(Transform(boom_text, stretch_text), FadeOut(micro_text), FadeOut(matrix_tex))
        self.wait(0.3)

        # Build suspense
        hype_text = Text("Time to *stretch* space 😤", font="Futura").scale(0.7).to_edge(UP)
        self.play(Transform(boom_text, hype_text), run_time=0.5)
        self.wait(0.3)

        # Matrix pop-up
        stretch_matrix_tex = MathTex(r"\begin{bmatrix} 2 & 0 \\ 0 & 1 \end{bmatrix}").scale(0.9).next_to(hype_text, DOWN).shift(LEFT)
        self.play(GrowFromCenter(stretch_matrix_tex), run_time=0.4)

        # Apply the stretch
        stretch_matrix = [[2, 0], [0, 1]]
        self.play(
            grid.animate.apply_matrix(stretch_matrix),
            cat.animate.apply_matrix(stretch_matrix),
            i_hat.animate.apply_matrix(stretch_matrix),
            j_hat.animate.apply_matrix(stretch_matrix),
            run_time=1
        )


        # Zoom in on the stretched cat
        self.camera.frame.save_state()
        self.play(self.camera.frame.animate.scale(1).move_to(cat), run_time=1)


        # Emphasize "stretch"
        boom_word = Text("S T R E T C H E D", font="Futura").scale(0.6).next_to(hype_text, DOWN, buff=1.2).shift(DOWN*0.3)
        fire = ImageMobject("Skull.png").scale(0.15).next_to(boom_word, RIGHT, buff=0.2)
        self.play(Write(boom_word), run_time=0.4)
        self.play(FadeIn(fire),run_time=0.4)
        self.wait(0.3)

        # Animate vector v → v′
        vector_v = Arrow(start=ORIGIN, end=[1, 0, 0], color=YELLOW)
        v_label = MathTex(r"\vec{v}").next_to(vector_v, DOWN)
        self.play(FadeIn(vector_v), FadeIn(v_label))

        # Show v becoming v′
        self.play(
            vector_v.animate.apply_matrix(stretch_matrix),
            v_label.animate.shift(RIGHT),  # Move label too
            run_time=0.7
        )
        v_prime_label = MathTex(r"\vec{v}'").move_to(v_label)
        self.play(Transform(v_label, v_prime_label))

        # Final math pop-up
        math_pop = MathTex(r"A \cdot \vec{v} = \vec{v}'").scale(0.8).to_edge(DOWN).shift(UP*2,RIGHT*2)
        self.play(FadeIn(math_pop), run_time=0.5)
        self.wait(1)

        # === UNSTRETCH SPACE BEFORE ROTATION ===
        self.wait(0.3)
        unstretch_text = Text("Let’s bring space back to normal...", font="Futura").scale(0.5).to_edge(UP)
        self.play(FadeOut(hype_text, boom_text,boom_word, fire, stretch_matrix_tex, math_pop, vector_v, v_label, v_prime_label), run_time=0.5)
        self.play(Write(unstretch_text))

        inverse_stretch = np.linalg.inv(stretch_matrix)
        self.play(
            grid.animate.apply_matrix(inverse_stretch),
            cat.animate.apply_matrix(inverse_stretch),
            i_hat.animate.apply_matrix(inverse_stretch),
            j_hat.animate.apply_matrix(inverse_stretch),
            run_time=1.5
        )

        self.play(FadeOut(unstretch_text))
        self.play(Restore(self.camera.frame), run_time=0.5)
        self.wait(0.3)



        # === ROTATION TRANSFORMATION ===
        self.wait(0.5)

        rotate_intro = Text("Want to rotate space?", font="Futura").scale(0.6).to_edge(UP)
        self.play(Write(rotate_intro))
        self.wait(0.3)

        # Rotation matrix definition
        rotation_angle = PI / 6  # 30 degrees
        rotate_matrix2 = [
            [np.cos(rotation_angle), -np.sin(rotation_angle)],
            [np.sin(rotation_angle), np.cos(rotation_angle)]
        ]

        matrix_tex2 = MathTex(
            r"\begin{bmatrix}"
            + f"{np.cos(rotation_angle):.2f} & { -np.sin(rotation_angle):.2f} \\\\"
            + f"{np.sin(rotation_angle):.2f} & {np.cos(rotation_angle):.2f}"
            + r"\end{bmatrix}"
        ).scale(0.9).next_to(rotate_intro, DOWN).shift(LEFT * 2)

        # Bounce the matrix in
        self.play(GrowFromCenter(matrix_tex2), run_time=0.4)
        self.wait(0.3)

        # Add a rotating vector with a trail
        trail_vector = Arrow(start=ORIGIN, end=[1, 0, 0], color=YELLOW)
        trail_dot = Dot(point=[1, 0, 0], color=YELLOW)
        self.play(FadeIn(trail_vector), FadeIn(trail_dot))

        # Animate vector rotating with a trail
        num_frames = 30
        trail = VGroup()
        for i in range(num_frames):
            angle = rotation_angle * (i + 1) / num_frames
            new_point = [np.cos(angle), np.sin(angle), 0]
            trail += Dot(point=new_point, radius=0.02, color=YELLOW)
        self.play(FadeIn(trail), run_time=0.5)

        # Apply the rotation matrix to everything
        self.play(
            grid.animate.apply_matrix(rotate_matrix2),
            cat.animate.apply_matrix(rotate_matrix2),
            i_hat.animate.apply_matrix(rotate_matrix2),
            j_hat.animate.apply_matrix(rotate_matrix2),
            trail_vector.animate.apply_matrix(rotate_matrix2),
            trail.animate.apply_matrix(rotate_matrix2),
            trail_dot.animate.apply_matrix(rotate_matrix2),
            run_time=2
        )

        # Tooltip: preserve length
        tip = Text("Rotation = same length, new direction", font="Futura").scale(0.4).next_to(matrix_tex2, DOWN).shift(LEFT*1.1)
        self.play(FadeIn(tip), run_time=0.5)

        # Outro matrix line
        matrix_label = Text("Yup, that’s a matrix too 😎", font="Futura").scale(0.5).next_to(tip, DOWN)
        self.play(FadeIn(matrix_label), run_time=0.5)
        self.wait(1)

                # === RESET SPACE AFTER ROTATION ===
        self.wait(0.5)
        self.play(FadeOut(rotate_intro, matrix_tex2, matrix_label, tip, trail_vector, trail, trail_dot))
        reset_text2 = Text("Okay, let’s reset space again...", font="Futura").scale(0.5).to_edge(UP)
        self.play(Write(reset_text2))

        # Inverse of the rotation matrix we just used
        inverse_rotate2 = np.linalg.inv(rotate_matrix2)

        self.play(
            grid.animate.apply_matrix(inverse_rotate2),
            cat.animate.apply_matrix(inverse_rotate2),
            i_hat.animate.apply_matrix(inverse_rotate2),
            j_hat.animate.apply_matrix(inverse_rotate2),
            run_time=1.5
        )

        self.play(FadeOut(reset_text2))
        self.wait(0.3)
        # === REFLECTION TRANSFORMATION (FINAL, CLEAN, WORKING) ===
        self.wait(0.3)

        reflect_intro = Text("Space, but mirrored 🔄", font="Futura").scale(0.6).to_edge(UP)
        self.play(Write(reflect_intro))

        # Define reflection matrix (across x-axis)
        reflect_matrix = [[1, 0], [0, -1]]

        # Show matrix and mirror line
        reflect_matrix_tex = MathTex(r"\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}").scale(0.9).next_to(reflect_intro, DOWN).shift(LEFT)
        mirror_line = DashedLine(start=LEFT * 7, end=RIGHT * 7, color=BLUE).shift(DOWN * 0.1)
        mirror_label = Text("mirror line", font="Futura").scale(0.4).next_to(mirror_line, DOWN, buff=0.1)

        self.play(FadeIn(reflect_matrix_tex), Create(mirror_line), FadeIn(mirror_label), run_time=0.6)
        self.wait(0.3)

        # Animate mirror line moving up slightly like it's cutting space
        self.play(
            mirror_line.animate.shift(UP * 0.2),
            mirror_label.animate.shift(UP * 0.2),
            run_time=0.4
        )

        # Then reflect everything EXCEPT the cat
        self.play(
            grid.animate.apply_matrix(reflect_matrix),
            i_hat.animate.apply_matrix(reflect_matrix),
            j_hat.animate.apply_matrix(reflect_matrix),
            cat.animate.flip(UP).move_to([cat.get_center()[0], -cat.get_center()[1], 0]),
            run_time=1.5
        )

        # Determinant = -1
        det_text = Text("Determinant = -1 → flips orientation", font="Futura").scale(0.45).next_to(reflect_matrix_tex, DOWN).shift(LEFT*0.6)
        self.play(FadeIn(det_text), run_time=0.5)

        # Double reflection comment
        bait_text = Text("Double reflection brings you back 👀", font="Futura").scale(0.45).next_to(det_text, DOWN).shift(DOWN * 0.3)
        self.play(FadeIn(bait_text))
        self.wait(0.4)

        # Reflect again to return to original
        self.play(
            grid.animate.apply_matrix(reflect_matrix),
            i_hat.animate.apply_matrix(reflect_matrix),
            j_hat.animate.apply_matrix(reflect_matrix),
            cat.animate.flip(UP).move_to([cat.get_center()[0], -cat.get_center()[1], 0]),
            run_time=1.5
        )
        self.wait(0.5)



        # === CLEAN UP REFLECTION ARTIFACTS ===
        self.play(
            FadeOut(reflect_intro),
            FadeOut(reflect_matrix_tex),
            FadeOut(mirror_line),
            FadeOut(mirror_label),
            FadeOut(det_text),
            FadeOut(bait_text),
            run_time=0.8
        )
        self.wait(0.3)



          # === FINAL SCENE: NON-INVERTIBLE MATRIX (TIKTOK VERSION) ===
        self.wait(0.5)

        # Tease something epic
        collapse_intro = Text("Now for the ultimate transformation...", font="Futura").scale(0.6).to_edge(UP)
        self.play(Write(collapse_intro))
        self.wait(0.5)

        # Show the "space destroyer" matrix
        collapse_matrix = [[1, 0], [0, 0]]
        collapse_tex = MathTex(r"\begin{bmatrix} 1 & 0 \\ 0 & 0 \end{bmatrix}").scale(0.9).next_to(collapse_intro, DOWN).shift(LEFT * 2)
        self.play(FadeIn(collapse_tex), run_time=0.5)
        self.wait(0.3)

        # Fake build-up line
        fake_hype = Text("This one’s gonna be beautiful ✨", font="Futura").scale(0.5).next_to(collapse_tex, DOWN).shift(RIGHT * 4.3,UP)
        self.play(FadeIn(fake_hype))
        self.wait(0.7)

        # Apply matrix to grid + cat
        self.play(
            grid.animate.apply_matrix(collapse_matrix),
            cat.animate.apply_matrix(collapse_matrix).stretch(0.01, dim=1),  # cartoon squish
            i_hat.animate.apply_matrix(collapse_matrix),
            j_hat.animate.apply_matrix(collapse_matrix),
            run_time=2
        )

        # Dramatic text after everything collapses
        collapse_text = Text("This one destroys space.", font="Futura").scale(0.6).to_edge(UP)
        self.play(Transform(collapse_intro, collapse_text), FadeOut(fake_hype))
        self.wait(0.5)

        # Add brainy detail
        rank_note = Text("Rank = 1 → flattened forever", font="Futura").scale(0.45).next_to(collapse_tex, DOWN).shift(RIGHT * 1.5)
        self.play(FadeIn(rank_note))
        self.wait(0.5)

        # Optional meme text
        tombstone = Text("the cat never recovered 💀", font="Futura").scale(0.4).next_to(rank_note, DOWN).shift(DOWN * 0.3)
        self.play(FadeIn(tombstone))
        self.wait(0.5)

        # Final CTA
        final_cta = Text("Follow for more math that bends space 📐", font="Futura").scale(0.5).to_edge(DOWN).shift(UP*3)
        self.play(FadeIn(final_cta), run_time=0.5)
        self.wait(1.5)
```
