```python
%%manim -pqh -r 1080,1920 WhatEvenIsIntegration

from manim import *
import numpy as np
Text.set_default(font="Avenir")

class WhatEvenIsIntegration(Scene):
    def construct(self):
        def add_highlight(mobj, color):
            return SurroundingRectangle(mobj, color=color, buff=0.2)

        def add_catcam():
            return ImageMobject("cat2.png").scale(0.4).to_corner(DR, buff=0.5)

        # Hook
        hook = Text("A-Level Maths: Brainrot Edition", font_size=54, color=RED)
        hook.scale(0.1).move_to(UP * 2)
        self.play(hook.animate.scale(10).move_to(ORIGIN), rate_func=there_and_back_with_pause, run_time=1.4)
        self.wait(0.3)

        subtitle = Text("Ep 5: What even *is* integration?", font_size=42, color=WHITE).next_to(hook, DOWN)
        self.play(FadeIn(subtitle, shift=UP), run_time=0.5)
        self.wait(1.3)
        self.play(FadeOut(hook), FadeOut(subtitle))

        # Catcam
        cat = add_catcam()
        self.add(cat)

        # Axes setup
        axes = Axes(
            x_range=[0, 5],
            y_range=[0, 5],
            axis_config={"include_tip": False}
        ).scale(1.2).move_to(DOWN * 0.5)
        labels = axes.get_axis_labels(x_label="x", y_label="y")
        self.play(Create(axes), FadeIn(labels))

        # Function: y = x^2
        curve = axes.plot(lambda x: x**2 / 5, color=YELLOW)
        curve_label = MathTex(r"y = \frac{1}{5}x^2").next_to(curve, RIGHT).shift(UP * 0.5)
        self.play(Create(curve), FadeIn(curve_label))

        # Area under the curve (0 to 4)
        area = axes.get_area(curve, x_range=[0, 4], color=BLUE, opacity=0.5)
        self.play(FadeIn(area))

        # Add explanation text
        area_text = Text("Integration = Area under the curve", font_size=34).to_edge(DOWN).shift(UP*3)
        self.play(Write(area_text), cat.animate.shift(LEFT*0.3))
        self.wait(1.5)

        self.play(FadeOut(area_text))

        # Animate slicing the area
        rects = axes.get_riemann_rectangles(curve, x_range=[0, 4], dx=0.5, color=GREEN, stroke_width=0.2)
        slice_text = Text("The area of infinite tiny rectangles.", font_size=30).to_edge(DOWN).shift(UP*3)
        self.play(Transform(area, rects), Write(slice_text), cat.animate.flip(RIGHT))
        self.wait(1.5)
        self.play(FadeOut(slice_text), FadeOut(rects))

        # Show integral symbol and solution
        integral = MathTex(r"\int_0^4 \frac{1}{5}x^2 \, dx = \frac{64}{15}", font_size=64, color=BLUE_C).to_edge(UP)
        int_text = Text("It's literally maths doing the summing for you", font_size=30).next_to(integral, DOWN)
        self.play(Write(integral), Write(int_text))
        self.wait(2)

        self.play(FadeOut(integral), FadeOut(int_text), FadeOut(curve_label))

        # Zoom out moment: show ∑ turning into ∫
        sum_to_int = MathTex(r"\sum \rightarrow \int", font_size=64, color=PURPLE).move_to(ORIGIN).shift(UP)
        zoom_text = Text("When sigma gets philosophical", font_size=32).next_to(sum_to_int, DOWN)
        self.play(Write(sum_to_int), Write(zoom_text), cat.animate.shift(UP*0.2))
        self.wait(1.8)

        self.play(FadeOut(sum_to_int), FadeOut(zoom_text), FadeOut(curve), FadeOut(area), FadeOut(axes), FadeOut(labels), FadeOut(cat))

        # Existential outro
        outro = Text("You thought it was just area...\nIt's literally reverse differentiation.", font_size=34, color=WHITE)
        self.play(Write(outro))
        self.wait(1.8)

        # Next ep tease
        tease = Text("Next Time: Definite vs indefinite integration.", font_size=40, color=RED).next_to(outro, DOWN)
        self.play(Write(tease))
        self.wait(2.5)
```
